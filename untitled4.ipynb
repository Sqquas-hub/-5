1. Сумма цифр трехзначного числа

x = abs(int(input("Введите трехзначное число:"))) — запрашивает ввод у пользователя, преобразует строку в целое число (int), а функция abs() берет его модуль (отбрасывает знак минус, если ввели отрицательное число, например -123 превратится в 123)

if 99 < x < 1000: — проверяет, действительно ли число является трехзначным (находится в диапазоне от 100 до 999 включительно)

y = x // 100 + x // 10 % 10 + x % 10 — разбивает число на отдельные цифры и складывает их:
	• x // 100 — находит первую цифру (сотни).
	• x // 10 % 10 — находит вторую цифру (десятки).
	• x % 10 — находит третью цифру (единицы).

print("Сумма цифр: ", y) — выводит результат на экран.

else: и print("Число не трехзначное") — срабатывают, если условие if не выполнилось (число меньше 100 или больше 999), и сообщают об ошибке.


```python
import math

x = abs(int(input("Введите трехзначное число:")))
if 99 < x < 1000:
    y = x // 100 + x // 10 % 10 + x % 10
    print("Сумма цифр: ", y)
else:
  print("Число не трехзначное")
```

```python
import math

def distance(x1, y1, x2, y2):
    return math.hypot(x2 - x1, y2 - y1)

x1 = float(input())
y1 = float(input())
x2 = float(input())
y2 = float(input())

print(distance(x1, y1, x2, y2))
```

```python
import math

def triangle_area(a, b, c):
    p = (a + b + c) / 2
    return math.sqrt(p * (p - a) * (p - b) * (p - c))

a = float(input())
b = float(input())
c = float(input())

print(triangle_area(a, b, c))
```
